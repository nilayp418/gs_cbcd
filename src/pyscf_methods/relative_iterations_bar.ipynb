# Inner-iteration counts: CG vs CBCD (Water vs MeOH)

Inner iterations for CG and CBCD ($c = N_o$), 6-31++G**. Bar width is proportional to $N_o$; height is the CG inner-iteration count, or the CBCD count divided by $N_o$, scaled so that the Water bars have the same height in both groups (labels give the unscaled values). Within each method, bar area is therefore proportional to the number of column updates (a CG step updates all $N_o$ columns, a CBCD step one). Dashed outline: estimated noninteracting supersystem (Table 2).

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import scipy as sp
from pyscf import gto, scf

# Locate rhf/code_in_c from the repo root (the dir containing manuscript/), so this works regardless of
# where the kernel was started.
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "ref_data").is_dir())
sys.path.insert(0, str(ROOT / "pyscf_methods" / "code_in_c"))
from fast_scf import cg_scf, cbcd_scf        # drop-ins for rhf/cg_scf.py, rhf/cbcd_scf.py (whole SCF in C)
from oracles import build_integrals, make_rhf_oracle_energy, make_rhf_grad_energy   # fast_scf puts ground_state/ on sys.path

mol1 = "Water"
mol2 = "MeOH"
ref_name = {mol1: "H2O", mol2: "CH3OH"}      # ref_data/<name>.txt
BASIS, REPS, EPS, DAMP = "6-31++G**", 101, 1e-6, 0.4

systems = {}
for mol in (mol1, mol2):
    with open(ROOT / "ref_data" / f"{ref_name[mol]}.txt") as fh:
        atom = "".join(fh.readlines()[1:])               # skip header row
    m = gto.M(atom=atom, basis=BASIS, verbose=0)
    ne = m.nelectron
    f = make_rhf_oracle_energy(build_integrals(m), ne)
    g0 = scf.RHF(m); g0.max_cycle = 0; g0.kernel()
    X = np.real(sp.linalg.fractional_matrix_power(g0.get_ovlp(), 0.5))
    systems[mol] = dict(f=f, grad=make_rhf_grad_energy(f), C0=X @ g0.mo_coeff[:, : ne // 2], no=ne // 2,
                        e_ref=scf.RHF(m).kernel() - m.get_enuc())
N_o = {mol: s["no"] for mol, s in systems.items()}

# timing=True returns (C, E, (wall, Fock-build time)); the inner-iteration time is wall - Fock-build time,
# as in pyscf_wrapper.ipynb. Median over REPS runs. CG: cg_scf; CBCD: cbcd_scf with c = N_o (Gauss-Southwell).
# Inner iteration counts come from one opt_profiling run (inner objective per outer iteration, start + each step).
def run(method, s):
    if method == "CG":
        return lambda **kw: cg_scf(s["f"], s["grad"], np.copy(s["C0"]), eps=EPS, damp_param=DAMP, **kw)
    return lambda **kw: cbcd_scf(s["f"], s["grad"], np.copy(s["C0"]), s["no"], eps=EPS, damp_param=DAMP, **kw)

t_inner, wall, iters = {"CG": {}, "CBCD": {}}, {"CG": {}, "CBCD": {}}, {"CG": {}, "CBCD": {}}
for key in ("CG", "CBCD"):
    for mol, s in systems.items():
        call = run(key, s)
        res = [call(timing=True) for _ in range(REPS)]       # (C, E, (wall, t_fock))
        assert all(abs(r[1] - s["e_ref"]) < 1e-4 for r in res), f"{key} {mol}: energy disagrees with PySCF"
        t_inner[key][mol] = float(np.median([r[2][0] - r[2][1] for r in res]))
        wall[key][mol] = float(np.median([r[2][0] for r in res]))
        _, _, (_, subs) = call(opt_profiling=True)
        iters[key][mol] = sum(len(x) - 1 for x in subs)
        print(f"{key:5s} {mol:5s} N_o={N_o[mol]}  inner time = {t_inner[key][mol] * 1e3:8.2f} ms  "
              f"total wall = {wall[key][mol] * 1e3:.2f} ms  inner iterations = {iters[key][mol]}")
```

```python
iters["CG"]["Water"] = 72
iters["CBCD"]["Water"] = 210
iters["CG"]["MeOH"] = 93
iters["CBCD"]["MeOH"] = 442
```

```python
from matplotlib.patches import Polygon, Patch

w_unit = 0.045                # bar width per occupied orbital (width ∝ N_o, so area ∝ column updates)
centers = {"CG": 0.0, "CBCD": 1.2}
colors = {mol1: "#4C72B0", mol2: "#DD8452"}
labels = {mol: rf"{name} ($N_o$ = {N_o[mol]})" for mol, name in [(mol1, r"H$_2$O"), (mol2, "MeOH")]}
pad_x, pad_y = 0.0, 0.0       # outline follows bar edges exactly
dash_kw = dict(fill=False, edgecolor="black", linestyle=(0, (3, 1.5)), linewidth=2.6, zorder=3)
dash_label = r"Noninteracting Supersystem (estimated)"

# Raw heights: CG inner iterations; CBCD inner iterations / N_o (a CG step updates N_o columns, a CBCD step one).
raw = {"CG": {mol: iters["CG"][mol] for mol in N_o},
       "CBCD": {mol: iters["CBCD"][mol] / N_o[mol] for mol in N_o}}
# Scale each method so its Water bar has unit height; areas compare within a method, not across methods.
height = {m: {mol: v / d[mol1] for mol, v in d.items()} for m, d in raw.items()}
ymax = max(max(d.values()) for d in height.values())

fig, ax = plt.subplots(figsize=(5.5, 4.2))
for method, xc in centers.items():
    h = height[method]
    wA, wB = w_unit * N_o[mol1], w_unit * N_o[mol2]
    left = xc - (wA + wB) / 2         # center the flush pair on xc
    for mol, x0, w in [(mol1, left, wA), (mol2, left + wA, wB)]:
        y = h[mol]
        ax.bar(x0, y, width=w, align="edge", color=colors[mol],
               edgecolor="black", linewidth=0.8,
               label=labels[mol] if method == "CG" else None)
        txt = f"{iters[method][mol]}" if method == "CG" else f"{raw[method][mol]:.1f}\n({iters[method][mol]})"
        ax.text(x0 + w / 2, y + 0.01 * ymax, txt,
                ha="center", va="bottom", fontsize=9, linespacing=0.95)

    # Dashed supersystem outline
    xl, xm, xr = left - pad_x, left + wA, left + wA + wB + pad_x
    y1, y2 = h[mol1] + pad_y, h[mol2] + pad_y
    if method == "CG":        # single box at the taller (MeOH) height
        verts = [(xl, 0), (xl, y2), (xr, y2), (xr, 0)]
    else:                     # step outline hugging each bar's height
        verts = [(xl, 0), (xl, y1), (xm, y1), (xm, y2), (xr, y2), (xr, 0)]
    ax.add_patch(Polygon(verts, closed=False, **dash_kw))

ax.set_xticks(list(centers.values()))
tick_labels = {"CG": "CG", "CBCD": r"CBCD, $c = N_o$"}
ax.set_xticklabels([tick_labels[m] for m in centers])
ax.set_yticks([])
ax.set_ylim(0, ymax * 1.15)
ax.spines[["top", "right", "left"]].set_visible(False)
handles, lbls = ax.get_legend_handles_labels()
handles.append(Patch(**{k: v for k, v in dash_kw.items() if k != "zorder"}))
lbls.append(dash_label)
ax.legend(handles, lbls, frameon=False, fontsize=9, ncol=2,
          loc="lower center", bbox_to_anchor=(0.5, 1.0),
          handlelength=1.6, handleheight=0.9, handletextpad=0.5,
          labelspacing=0.35, columnspacing=1.2, borderaxespad=0.2)
fig.tight_layout()
fig.savefig("relative_iterations_bar.pdf", bbox_inches="tight")
fig.savefig("relative_iterations_bar.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# TOC graphic: same plot redrawn at achemso's TOC box size (3.25 x 1.75 in), legend moved to the right.
# Saved at exactly that size (no bbox_inches="tight") so it fills the tocentry box without rescaling.
toc_centers = {"CG": 0.0, "CBCD": 0.85}
toc_dash_kw = {**dash_kw, "linewidth": 1.6}

fig_toc, ax = plt.subplots(figsize=(3.25, 1.75))
fig_toc.subplots_adjust(left=0.02, right=0.64, bottom=0.14, top=0.97)
for method, xc in toc_centers.items():
    h = height[method]
    wA, wB = w_unit * N_o[mol1], w_unit * N_o[mol2]
    left = xc - (wA + wB) / 2
    for mol, x0, w in [(mol1, left, wA), (mol2, left + wA, wB)]:
        y = h[mol]
        ax.bar(x0, y, width=w, align="edge", color=colors[mol], edgecolor="black", linewidth=0.6,
               label=labels[mol] if method == "CG" else None)
        txt = f"{iters[method][mol]}" if method == "CG" else f"{raw[method][mol]:.1f}\n({iters[method][mol]})"
        ax.text(x0 + w / 2, y + 0.01 * ymax, txt, ha="center", va="bottom", fontsize=6.5, linespacing=0.9)
    xl, xm, xr = left, left + wA, left + wA + wB
    y1, y2 = h[mol1], h[mol2]
    if method == "CG":
        verts = [(xl, 0), (xl, y2), (xr, y2), (xr, 0)]
    else:
        verts = [(xl, 0), (xl, y1), (xm, y1), (xm, y2), (xr, y2), (xr, 0)]
    ax.add_patch(Polygon(verts, closed=False, **toc_dash_kw))

ax.set_xticks(list(toc_centers.values()))
ax.set_xticklabels([tick_labels[m] for m in toc_centers], fontsize=8)
ax.tick_params(axis="x", length=2, pad=1.5)
ax.set_yticks([])
ax.set_ylim(0, ymax * 1.18)
ax.spines[["top", "right", "left"]].set_visible(False)
ax.spines["bottom"].set_linewidth(0.8)
handles, lbls = ax.get_legend_handles_labels()
handles.append(Patch(**{k: v for k, v in toc_dash_kw.items() if k != "zorder"}))
lbls.append("Noninteracting\nsupersystem")
ax.legend(handles, lbls, frameon=False, fontsize=7.5, loc="center left", bbox_to_anchor=(1.0, 0.5),
          handlelength=1.4, handleheight=1.0, handletextpad=0.5, labelspacing=0.8, borderaxespad=0.3)
fig_toc.savefig(FIGDIR / "relative_iterations_bar_toc.pdf")
fig_toc.savefig(FIGDIR / "relative_iterations_bar_toc.png", dpi=600)
plt.show()
```
